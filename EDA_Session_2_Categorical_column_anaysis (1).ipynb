**bar Chart**

- X-axis: class names(keys)

- y-axis: Frequency (values)

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
```

```python
plt.figure(figsize=(10,4))
plt.bar(keys,values)
plt.xlabel('Continents')
plt.ylabel('No Of applicants')
plt.title('Bar chart')
plt.show()
```

```python
from matplotlib.colors import CSS4_COLORS
CSS4_COLORS
```

```python
colors=['Red','Blue','Green','Yellow','Pink','Purple']
plt.figure(figsize=(10,4))
plt.bar(keys,
        values,
        color=colors)
plt.xlabel('Continents')
plt.ylabel('No Of applicants')
plt.title('Bar chart')
plt.savefig('Continent_bar.jpg')
plt.show()
```

**pie chart**

- autopct= %a.bf%%

- first % is the formatting

- a : how much min width

- b : after point how many decimals

- %% : % symbol

- %1.1f%%: 40.2%

- %2.2f%% :423.23

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
#plt.pie(x,labels)
plt.pie(values,
       explode=[0.1 for i in range(6)],
       labels=keys,
       autopct='%1.2f%%',
       radius=0.8,
       startangle=40)
plt.savefig('continent_pie.jpg')
plt.show()
```
